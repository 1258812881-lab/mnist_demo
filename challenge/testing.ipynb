```python
import numpy as np
import joblib  
from sklearn.metrics import accuracy_score
```

```python
svc_model=joblib.load('../models/svm_mnist_model.pkl')  #导入训练好的模型
X = np.loadtxt('cdigits_digits_vec.txt', delimiter='\t')
y=np.loadtxt('cdigits_digits_labels.txt', delimiter='\t')
X_norm = X / 255.0
X.shape
```

```python
y_pred = svc_model.predict(X_norm)
y_pred
```

```python
accuracy = accuracy_score(y, y_pred)
accuracy
```
