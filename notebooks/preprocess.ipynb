```python
import numpy as np 
import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import matplotlib
from sklearn.decomposition import PCA
```

```python
X = np.loadtxt("../data/raw/digits4000/digits4000_digits_vec.txt", delimiter="\t")
y = np.loadtxt("../data/raw/digits4000/digits4000_digits_labels.txt", delimiter="\t")
X.shape
```

```python
#可视化
%matplotlib inline
some_digit= X[1].reshape(28,28)
plt.imshow(some_digit,cmap=matplotlib.cm.binary,interpolation="nearest")
plt.axis("off")
plt.show()
```

```python
y[1]
```

```python
#归一化，将像素值从[0,255]缩放到[0,1]
X_minmax = X / 255.0
X_minmax.shape
```

```python
some_digit= X_minmax[3600].reshape(28,28)
plt.imshow(some_digit,cmap=matplotlib.cm.binary,interpolation="nearest")
plt.axis("off")
plt.show()
```

```python
#主成分分析
pca= PCA(n_components=0.95) #保留方差比为95%的主成分
X_pca = pca.fit_transform(X_minmax)
X_pca.shape #最终将784个特征保留到147个
```

```python
#保存处理后的数据
np.savetxt('../data/processed/digits4000_normalized.txt', X_minmax, delimiter='\t', fmt='%.6f')
np.savetxt('../data/processed/digits4000_pca.txt', X_pca, delimiter='\t', fmt='%.6f')
```
